```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("expenses.csv")
data
```

```python
#1. **Total amount spent** (add up all the amounts)

Total_spend_money= data.Amount.sum()
print(Total_spend_money)

#2. **Average spending per transaction**

Average_transaction_spend = Total_spend_money / len(data)
print(Average_transaction_spend)

# 3. **How much you spent in each category** (Food, Transport, etc.)

category_data =(data.groupby("Category")['Amount'].sum())
print(category_data)

# 4. **Which category you spent THE MOST money on
max_spend = data.groupby('Category')['Amount'].sum()
sort_values = max_spend.sort_values(ascending=False)
max_spend_category = sort_values.index[0]
print("I spent most of the money on", max_spend_category, "amount is", sort_values.iloc[0])
```

```python
# Visulaization

x_axis = category_data.index
y_axis = category_data.values

plt.bar(x_axis,y_axis)
plt.xlabel("Category names")
plt.ylabel("Amount spent in dollars")
plt.title("Spend Category")
plt.show()
```

```python
#1. Calculate total amount spent across all expenses
Total_spend_money = data['Amount'].sum()
print(f"1.💰 Total Money spend is ${Total_spend_money:.2f}")
print("-" * 40)

#2. **Average spending per transaction**
Average_transaction_spend = Total_spend_money / len(data)
print(f"2.💰 Average Trasaction spend is ${Average_transaction_spend:.2f}")
print("-" * 40)

#3. # 3. **How much you spent in each category** (Food, Transport, etc.)
category_data =(data.groupby("Category")['Amount'].sum())
print(f'3.📊 Spending amount per category is below')
for names,values in category_data.items():
  print(f'{names} : $ {values:.2f}')
print("-" * 40)

# 4. **Which category you spent THE MOST money on
max_spend = data.groupby('Category')['Amount'].sum()
sort_values = max_spend.sort_values(ascending=False)
max_spend_category = sort_values.index[0]
print(f"4.Spend most of the money on",max_spend_category, "and amount is", sort_values.iloc[0])
print("-" * 40)
print(f"5.Visualization of Spend Analysis")
x_axis = category_data.index
y_axis = category_data.values
for names,values in category_data.items():
  plt.text(names,values, f'{values:.2f}',ha='center',va='bottom')
colors = ['Blue','Orange','Red','Pink','Green']
plt.bar(x_axis,y_axis,color=colors)
plt.xlabel("Category names")
plt.ylabel("Amount spent in dollars")
plt.title("Spend Category")
plt.figure(figsize=(10,6))
plt.savefig('Spend Analysis.png')
plt.show()
```
